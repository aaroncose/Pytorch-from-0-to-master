```python
### CHAPTER 1 - MODEL FROM SCRATCH

##SIMPLE LINEAL REGRESION ---> y = XW + b

# X = input data
# W = weight ----> knob
# b = bias   ----> knob
# y(y-hat) = our model´s prediction (our guess)

# W and b are the the "tools" or "knobs" our model can turn.
# Our goal is to find the perfect values fot them to  prediction y-hat as close as the real y as possible
```

# 5 STEP LOOP

```python
## STEP 1 Create Data

#Our batch of data will have 10 data points
N = 10
#Each data point has 1 input feature and 1 output value
D_in = 1
D_out = 1

#Create our input data X
X = torch.randn(N, D_in)

#Create our true target labels y by using the "true" W and b
#The "true" W is 2.0, the "true" b is 1.0
true_W = torch.tensor([[2.0]])
true_b = torch.tensor(1.0)
y_true = X @ true_W + true_b + torch.randn(N, D_out) * 0.1

#The model will never see the **true_W OR true_b**, its job is discover them just by looking at X AND y_true

print(y_true)
```

```python
# MODEL´S BRAIN

# Initialize our parameters with random values.
# Shapes must be correct for matrix multiplication!

W = torch.randn(D_in, D_out, requires_grad=True)
b = torch.randn(1, requires_grad=True)

print(f"Initial Weight W:\n {W}")
print(f"Initial bias b:\n {b}")
```

```python
## IMPLEMNTATION FROM MATH TO CODE ##

# The math
# y = XW + b

# The code
# y_hat = X @ W + b

# Define model´s logic
def model(datos_entrada):
    return datos_entrada @ W + b
```

```python
#Forward pass through our model
y_hat = model(X)

print(f"Prediction y_hat (first 3 rows):\n {y_hat[:3]}\n")
print(f"True Labels y_true( first 3 rows):\n {y_true[:3]}")
```

```python
## STEP 2 Quantify the error

# Mean Squared Error (MSE)
# loss = ((y_hat - y_true)**2).mean()

# Mean Squared Error (MSE)

# $$L = \frac{1}{N} \sum_{i=1}^{N} (\hat{y}_i - y_i)^2$$

#   1. Difference: Subtract the real value from the prediction: (y_hat - y)
#   2. Square: Square the result to make it positive: (y_hat - y)^2
#   3. Average: Calculate the mean of all squared differences: .mean()

#   Remember the "true" W is 2.0, the "true" b is 1.0


# Lets calculate the loss
error = y_hat - y_true
squared_error = error ** 2
loss = squared_error.mean()

print(f"Loss (Our single scorecard number): {loss}")
```

```python
# Compute grradients
loss.backward()

#The gradients are now stored in the .grad attribute
print(f"Gradient for W (aL/aW):\n {W.grad}\n")
print(f"Gradient for b (aL/ab):\n {b.grad}\n")
```



$$\theta_{t+1} = \theta_t - \eta \cdot \nabla_\theta L$$

* **$\theta$ (theta):** Represents all our parameters. In our case, these are the "knobs" **W** and **b**.
* **$\eta$ (eta):** This is the **learning rate**. It is a small number (e.g., 0.01) that controls our step size.
* **$\nabla_\theta L$:** This is the gradient of the loss. We just calculated this using `.backward()`, and it is stored in `W.grad` and `b.grad`.

```python
# Updating the weights (W)
# W_new = W_old - learning_rate * W.grad

# Updating the bias (b)
# b_new = b_old - learning_rate * b.grad
```

```python
# Hyperparameters
learning_rate, epochs = 0.01, 100

# Re-initialize parameters
W, b = torch.randn(1, 1, requires_grad=True), torch.randn(1, requires_grad=True)

# Training Loop
for epoch in range(epochs):
    # Forward pass and loss
    y_hat = X @ W + b
    loss = torch.mean((y_hat - y_true)**2)

    # Backward pass
    loss.backward()

    # Update parameters
    with torch.no_grad():
        W -= learning_rate * W.grad
        b -= learning_rate * b.grad

    # Zero gradients
    W.grad.zero_()
    b.grad.zero_()

    # ... (dentro del bucle/loop)
    if epoch % 10 == 0:
        print(f"Epoch {epoch:02d}: Loss={loss.item():.4f}, W={W.item():.3f}, b={b.item():.3f}")

print(f"\nFinal Parameters: W={W.item():.3f}, b={b.item():.3f}")
print(f"True Parameters:  W=2.000, b=1.000")
```
